```python
#read N number from the console list develop a program to print mean variance and standard deviation with suitable message
x=int(input("enter the element of list"))
l=[]
for i in  range(x):
    l.append(int(input("enter the numbers")))
print(l)
mean=sum(l)/x
v=0
for i in l:
    v=v+(i-mean)**2
var=v/x
import math
sd=math.sqrt(var)
print("mean=",mean,"varience=",var,"standard deviation=",sd)
```

```python
x=int(input("enter the element of list"))
l=[]
for i in  range(x):
    l.append(int(input("enter the numbers")))
print(l)
mean=sum(l)/x
v=0
for i in l:
    v=v+(i-mean)**2
var=v/x
import math
sd=math.sqrt(var)
print(f"mean={mean:0.2f},varience={var:0.2f},deviation={sd:0.2f}")
```
